```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from numpy.random import rand
```

```python
# train test split logic
data = pd.read_csv('Housing.csv')
data = data.loc[:,['price','area','bedrooms','bathrooms','stories','parking']]
new_data = data.sample(545,ignore_index=True,random_state=2)
test = new_data.iloc[:109,:].sample(109,ignore_index=True,random_state=42)
train = new_data.iloc[109:,:].sample(436,ignore_index=True,random_state=42)
X_train = train[['area','bedrooms','bathrooms','stories','parking']]
X_test= test[['area','bedrooms','bathrooms','stories','parking']]
y_train = train['price']
y_test = test['price']
```

```python
# mean and standard deviation on X_train's area column
u = X_train['area'].mean()
n = X_train['area'].size
variance = (((X_train['area']-u)**2).sum())/n
sigma = (variance)**0.5
sigma
```

```python
  # Z score logic
# z = (x-u)/sigma
X_train['area'] = (X_train['area']-u)/sigma
```

```python
# Z score on test col but using test u and sigma
X_test['area'] = (X_test['area'] - u)/sigma
```

```python
# Scaling our y
# mean and standard deviation on y_train's area column
u = y_train.mean()
n = y_train.size
variance = (((y_train-u)**2).sum())/n
sigma = (variance)**0.5
sigma
```

```python
# scaling y column
y_train = (y_train-u)/sigma
y_test = (y_test-u)/sigma
y_train
```

```python
# gradient desent on this dataset
no_of_col = X_train.shape[1]
w = np.zeros(no_of_col)
b = 0
m = X_train.shape[0]
def derivative_w(w,b):
   r = ((X_train @ w + b) - y_train)@X_train
   return r/m
def derivative_b(w,b):
    r =((X_train @ w + b )- y_train)
    r = r.sum()
    return r/m
def gradient_descent(w,b):
    alpha = 0.001;
    max_itr = 50000
    cond = 0.0000001
    for _ in range(max_itr):
      temp_w = w - alpha*derivative_w(w,b)
      temp_b = b - alpha*derivative_b(w,b)
      # euclidean distance
      w_change = ((temp_w - w) ** 2).sum() ** 0.5
      b_change = abs(temp_b - b)
      if w_change < cond and b_change < cond:
          break
      else:
        b = temp_b
        w = temp_w
    return w,b
```

```python
w,b = gradient_descent(w,b);
w,b
```

```python
# r2 score on test data
y_pred = X_test@w + b
SSr = ((y_test-y_pred )**2).sum()
SSm = ((y_test - y_test.mean())**2).sum()
r2_test = 1 - (SSr/SSm)
r2_test
```

```python
y_pred = X_train@w + b
SSr = ((y_train-y_pred )**2).sum()
SSm = ((y_train - y_train.mean())**2).sum()
r2_train = 1 - (SSr/SSm)
r2_train
```

```python
# adjusted r2 score on train
n = X_train.shape[0]
k = X_train.shape[1]
r2_adj_train = 1 - ((1-r2_train)*(n-1))/(n-1-k)
r2_adj_train
```

```python
# adjusted r2 score on test
n = X_test.shape[0]
k = X_test.shape[1]
r2_adj_test = 1 - ((1-r2_test)*(n-1))/(n-1-k)
r2_adj_test
```
