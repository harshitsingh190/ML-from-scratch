```python
import math
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('diabetes.csv')
new_data = data.sample(768,ignore_index=True,random_state=2)
test = new_data.iloc[:154,:].sample(154,ignore_index=True,random_state=42)
train = new_data.iloc[154:,:].sample(614,ignore_index=True,random_state=42)
X_train = train.iloc[:,0:8]
X_test= test.iloc[:,0:8]
y_train = train.iloc[:,-1]
y_test = test.iloc[:,-1]
```

```python
u = X_train.mean()
st = X_train.std()
X_train_scaled = (X_train-u)/st
X_test_scaled = (X_test-u)/st
X_train_scaled
```

```python
no_of_col = X_train.shape[1]
w = np.zeros(no_of_col)
b = 0
m = X_train.shape[0]

def derivative_j_w(w,b):
    f = 1/(1+np.exp(-(X_train_scaled@w+b)))
    r = (f-y_train)@X_train_scaled
    return r/m

def derivative_j_b(w,b):
    f = 1/(1+np.exp(-(X_train_scaled@w+b)))
    r = (f-y_train).sum()
    return r/m

def gradient_decent(w,b):
    alpha = 0.01
    max_itr = 50000
    cond = 0.0000001
    for _ in range(max_itr):
        temp_w = w - alpha*derivative_j_w(w,b)
        temp_b = b - alpha*derivative_j_b(w,b)
        # euclidean distance
        w_change = ((temp_w - w) ** 2).sum() ** 0.5
        b_change = abs(temp_b - b)
        if w_change < cond and b_change < cond:
          break
        else:
            b = temp_b
            w = temp_w
    return w,b

a,b = gradient_decent(w,b)
print(a," " ,b)
```

```python
f = 1/(1+np.exp(-(X_test_scaled@a+b)))
y_pred = (f >= 0.5).astype(int)
accuracy = (y_test==y_pred).mean()
print(accuracy)
```
